```python
import pandas as pd
import numpy as np

try:
    df = pd.read_csv('calon_tni.csv')
except FileNotFoundError:
    print("File 'calon_tni.csv' tidak ditemukan. Pastikan file berada di direktori yang benar.")
    exit()

# Menampilkan data awal
print("=== Data Calon TNI ===")
print(df)
print("\n" + "="*30 + "\n")

print("=== Statistik Dasar Tinggi Badan ===")
statistik_dasar = df['tinggi'].describe()
print(statistik_dasar)
print("\n" + "="*30 + "\n")

rata_rata_tinggi = df['tinggi'].mean()
tertinggi = df['tinggi'].max()
terendah = df['tinggi'].min()
median_tinggi = df['tinggi'].median()

print(f"Rata-rata tinggi : {rata_rata_tinggi:.2f} cm")
print(f"Tertinggi          : {tertinggi} cm")
print(f"Terendah           : {terendah} cm")
print(f"Median             : {median_tinggi} cm")
print("\n" + "="*30 + "\n")

print("=== Calon di atas rata-rata tinggi ===")
di_atas_rata_rata = df[df['tinggi'] > rata_rata_tinggi]
print(di_atas_rata_rata)
print("\n" + "="*30 + "\n")

print("=== Urutan dari Tertinggi ke Terendah ===")
urutan_tertinggi = df.sort_values(by='tinggi', ascending=False)
print(urutan_tertinggi)
print("\n" + "="*30 + "\n")

bins = [0, 169, 179, float('inf')]
labels = ['Pendek', 'Sedang', 'Tinggi']
df['kategori'] = pd.cut(df['tinggi'], bins=bins, labels=labels, right=True)

print("=== Data dengan Kategori ===")
print(df)
print("\n" + "="*30 + "\n")

# Menghitung jumlah calon per kategori
print("=== Jumlah Calon per Kategori ===")
jumlah_per_kategori = df['kategori'].value_counts()
print(jumlah_per_kategori)
print("\n" + "="*30 + "\n")

std_lanjutan = np.std(df['tinggi'])
var_lanjutan = np.var(df['tinggi'])

print("=== Statistik Lanjutan ===")
print(f"Standar Deviasi: {std_lanjutan:.2f}")
print(f"Variansi         : {var_lanjutan:.2f}")
print("\n" + "="*30 + "\n")

print("=== Calon dengan Tinggi Badan Tertinggi ===")
calon_tertinggi = df.loc[df['tinggi'].idxmax()]
print(f"Nama   : {calon_tertinggi['nama']}")
print(f"Alamat : {calon_tertinggi['alamat']}")
print(f"Tinggi : {calon_tertinggi['tinggi']} cm")
```
